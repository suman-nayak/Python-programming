```python
#1. Write a program to find the smallest and largest digit in a number.
```

```python
n = int(input("Enter a number: "))
smallest = 9
largest = 0
while n > 0:
    digit = n % 10

    if digit < smallest:
        smallest = digit
        
    if digit > largest:
        largest = digit
        
    n = n // 10

print("Smallest digit =", smallest)
print("Largest digit =", largest)
```

```python
#2. Write a program to find the frequency of each digit in a number.
```

```python
n = int(input("Enter a number: "))
for i in range(10):
    count = 0
    temp = n
    while temp > 0:
        digit = temp % 10
        if digit == i:
            count = count + 1

        temp = temp // 10

    if count > 0:
        print(i, "=", count, "times")
```

```python
#3. Write a program to check whether a number is binary or not.
```

```python
n = int(input("Enter a number: "))

temp = n
binary = True

while temp > 0:
    digit = temp % 10

    if digit != 0 and digit != 1:
        binary = False
        break

    temp = temp // 10

if binary:
    print("Binary number")
else:
    print("Not a binary number")
```

```python
#Write a program to count how many digits are even and odd in a number.
```

```python
n = int(input("Enter a number: "))
even = 0
odd = 0
while n > 0:
    digit = n % 10

    if digit % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

    n = n // 10

print("Number of even digits =", even)
print("Number of odd digits =", odd)
```

```python
#8. Print numbers from 1 to N, but skip numbers whose digit sum is divisible by 3.
```

```python
n = int(input("Enter N: "))

for i in range(1, n + 1):
    temp = i
    sum = 0

    while temp > 0:
        digit = temp % 10
        sum = sum + digit
        temp = temp // 10

    if sum % 3 == 0:
        continue
    else:
        print(i)
```

```python
#9.Count numbers between 1 and 100 whose reverse is greater than the original number.
```

```python
count = 0
for i in range(1, 101):
    temp = i
    reverse = 0

    while temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp = temp // 10

    if reverse > i:
        count = count + 1

print("Count =", count)
```

```python
#10.Print all numbers between 1 and 100 where product of digits > sum of digits.
```

```python
for i in range(1, 101):
    temp = i
    sum = 0
    product = 1

    while temp > 0:
        digit = temp % 10

        sum = sum + digit
        product = product * digit

        temp = temp // 10

    if product > sum:
        print(i)
```

```python
#11. Print all numbers from 1 to 100 that are not divisible by any digit they contain.
```

```python
for i in range(1, 101):
    temp = i
    divisible = False

    while temp > 0:
        digit = temp % 10
        temp = temp // 10
        
        if digit == 0:
            continue

        if i % digit == 0:
            divisible = True
            break

    if divisible == False:
        print(i)
```

```python
#13. Swap the first and last digit of a number.
```

```python
n = int(input("Enter a number: "))
temp = n
last = n % 10
div = 1

while temp >= 10:
    temp = temp // 10
    div = div * 10

first = temp
middle = (n % div) // 10
print("Number after swapping =", last * div + middle * 10 + first)
```

```python
#14. Count how many numbers between 1 and 100 contain exactly two digits same.
```

```python
count = 0

for i in range(1, 101):
    temp = i
    digit_count = 0
    first_digit = -1

    while temp > 0:
        digit = temp % 10

        if first_digit == -1:
            first_digit = digit
        elif digit == first_digit:
            digit_count = digit_count + 1

        temp = temp // 10

    if digit_count == 1:
        count = count + 1

print("Count =", count)
```

```python
#15. Check whether a number is a Harshad number (divisible by sum of digits).
```

```python
n = int(input("Enter a number: "))

temp = n
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + digit
    temp = temp // 10

if n % sum == 0:
    print("Harshad number")
else:
    print("Not a Harshad number")
```

```python
#16. Find numbers whose square ends with the same digits as the number itself.(Hit: 5->5^2=25)
```

```python
#17. Find numbers between 1 and 1000 that become prime after removing one digit.
```

```python
for n in range(10, 1001):

    temp = n
    digits = 0

    while temp > 0:
        digits = digits + 1
        temp = temp // 10

    divisor = 1
    for i in range(1, digits):
        divisor = divisor * 10

    temp = n
    found = False

    while divisor > 0:

        left = n // (divisor * 10)
        right = n % divisor

        new_number = left * divisor + right

        if new_number < 2:
            divisor = divisor // 10
            continue

        prime = True
        j = 2

        while j * j <= new_number:
            if new_number % j == 0:
                prime = False
                break
            j = j + 1

        if prime:
            found = True
            break

        divisor = divisor // 10

    if found:
        print(n)
```

```python
#18. Print all numbers from 1 to 500 that are divisible by every non-zero digit they contain.
```

```python
for n in range(1, 501):

    temp = n
    valid = True

    while temp > 0:
        digit = temp % 10
        temp = temp // 10

        if digit == 0:
            continue

        if n % digit != 0:
            valid = False
            break

    if valid:
        print(n)
```

```python
#20. Find the LCM and GCD of two numbers.
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
x = a
y = b

while y != 0:
    remainder = x % y
    x = y
    y = remainder

gcd = x
lcm = (a * b) // gcd

print("GCD =", gcd)
print("LCM =", lcm)
```
