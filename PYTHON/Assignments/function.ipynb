# lambda function

syntax: lambda input: Expression

```python
a = lambda x,y: x + y
print("Output:",a(5,5))
```

```python
max = lambda x,y: x if x > y else y
print(max(1,2))
```

```python
OE = lambda x: "even"if x % 2 == 0 else "Odd"
print(OE(5))
print(OE(6))
```

# lambdas in filter()

Syntax: filter(function, sequence)

```python
lst = [10, 2, 3, 8, 7, 5, 4, 3, 11, 0, 1]
filtered_result = filter(lambda x: x > 4, tuple(lst))
#print(list(filtered_result))
print(tuple(filtered_result))
```

```python
def check(x):
    if x >= 10:
        return True
    else:
        return False

lst = [10, 2, 3, 27, 32, 29]
result = filter(check, lst)
print(list(result))
```

```python
number = [10, 15, 20, 15, 30, 35, 40]
result = filter(lambda x: x % 2 == 0, number)
print(list(result))
```

# lambda with map()

Syntax: map(function, sequence)

```python
lst = [1, 2, 3, 4, 5]
lsti = list(map(lambda x:x*x, lst))
print(lsti)
```

```python
l1 = [1, 2, 3, 4, 5]
sq = map(lambda x: x**2, filter(lambda x: x % 2 == 0, l1))

print(list(sq))
```
