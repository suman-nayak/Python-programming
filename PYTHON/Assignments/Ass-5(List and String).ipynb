1. Given a range of first 10 numbers, Iterate from start number to the end number and print the 
sum of the current number and previous number. 

```python
start = 1
end = 10

previous = start

for i in range(start, end + 1):
    if i == start:
        print(i)
    else:
        print(i + previous)
        previous = i
```

18. Arrange string characters such that lowercase letters should come first.

19. Count all lower case, upper case, digits, and special symbols from a given string. 

```python
s = input("Enter a string: ")

lower = 0
upper = 0
digit = 0
special = 0

for i in s:
    if i.islower():
        lower = lower + 1
    elif i.isupper():
        upper = upper + 1
    elif i.isdigit():
        digit = digit + 1
    else:
        special = special + 1

print("Lowercase:", lower)
print("Uppercase:", upper)
print("Digits:", digit)
print("Special symbols:", special)
```

20. Given a string, return the sum and average of the digits that appear in the string, ignoring all 
other characters. 

```python
s = input("Enter a string: ")

sum = 0
count = 0

for i in s:
    if i.isdigit():
        sum = sum + int(i)
        count = count + 1

if count > 0:
    average = sum / count
else:
    average = 0

print("Sum =", sum)
print("Average =", average)
```

23. Write a Python program to sort a list of elements using the bubble sort algorithm. 

```python
numbers = [5, 2, 8, 1, 3]

n = len(numbers)

for i in range(n):
    for j in range(0, n - i - 1):
        if numbers[j] > numbers[j + 1]:
            temp = numbers[j]
            numbers[j] = numbers[j + 1]
            numbers[j + 1] = temp

print("Sorted list:", numbers)
```

25. Write a Python program to count the frequency of each element in a list. 

```python
numbers = [10, 20, 10, 30, 20, 10, 40]

checked = []

for i in numbers:
    if i not in checked:
        count = 0

        for j in numbers:
            if i == j:
                count = count + 1

        print(i, "=", count)
        checked.append(i)
```

30. Write a Python Program to print all Possible Combinations from the three Digits. 

```python
d1, d2, d3 = 1, 2, 3
digits = [d1, d2, d3]

print("Possible combinations:")
for i in range(3):
    for j in range(3):
        for k in range(3):
            # Ensure each digit is used only once per combination
            if i != j and j != k and i != k:
                print(digits[i], digits[j], digits[k])
```
