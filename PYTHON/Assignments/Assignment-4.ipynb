1. Given a string of odd length greater 7, return a string made of the middle three chars of a given 
String.

```python
s = input("Enter a string: ")

if len(s) > 7 and len(s) % 2 != 0:
    mid = len(s) // 2
    print("Middle three characters:", s[mid - 1:mid + 2])
else:
    print("Invalid input")
```

2. Given 2 strings, s1 and s2, create a new string by appending s2 in the middle of s1.

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

mid = len(s1) // 2

result = s1[:mid] + s2 + s1[mid:]

print("New string:", result)
```

3. Given 2 strings, s1, and s2 return a new string made of the first, middle and last character of 
each input string.

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

mid1 = len(s1) // 2
mid2 = len(s2) // 2

result = s1[0] + s1[mid1] + s1[-1] + s2[0] + s2[mid2] + s2[-1]

print("New string:", result)
```

4. Arrange string characters such that lowercase letters should come first.  

```python
s = input("Enter a string: ")
lower = ""
others = ""

for ch in s:
    if ch.islower():
        lower += ch
    else:
        others += ch

result = lower + others
print("Result:", result)
```

5. Count all lower case, upper case, digits, and special symbols from a given string.

```python
s = input("Enter a string: ")
lower = 0
upper = 0
digit = 0
special = 0

for ch in s:
    if ch.islower():
        lower += 1
    elif ch.isupper():
        upper += 1
    elif ch.isdigit():
        digit += 1
    else:
        special += 1

print("Lowercase letters:", lower)
print("Uppercase letters:", upper)
print("Digits:", digit)
print("Special symbols:", special)
```

6. Given a string, display the sum and average of the digits that appear in the string, ignoring all 
other characters.

```python
s = input("Enter a string: ")
sum = 0
count = 0

for ch in s:
    if ch.isdigit():
        sum += int(ch)
        count += 1

if count > 0:
    average = sum / count
    print("Sum =", sum)
    print("Average =", average)
else:
    print("No digits found")
```

7. Given a string, display only those characters which are present at an even index number. 

```python
s = input("Enter a string: ")
result = ""

for i in range(len(s)):
    if i % 2 == 0:
        result += s[i]

print("Characters at even indexes:", result)
```

8. Given a string and an integer number n, remove characters from a string starting from zero up 
to n and return a new string.

```python
s = input("Enter a string: ")
n = int(input("Enter n: "))

if n >= 0 and n < len(s):
    result = s[n + 1:]
    print("New string:", result)
else:
    print("Invalid value of n")
```

9. Return the total count of sub-string appears in the given string. 

```python
s = input("Enter a string: ")
sub = input("Enter the substring: ")
count = 0
start = 0

while True:
    pos = s.find(sub, start)

    if pos == -1:
        break

    count += 1
    start = pos + 1

print("Total count:", count)
```
