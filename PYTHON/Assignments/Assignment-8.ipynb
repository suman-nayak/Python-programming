1. Write a function that returns the    sum of digits    of a number using recursion. 

```python
def sod(num):
    if num == 0:
        return 0
    else:
        return  num%10+sod(num//10)
    
n = int(input("Enter a number: "))  
print("Sum is:",sod(n))
```

2. Write a function that checks whether a number is an    Armstrong number    (without using built
in power function).

```python
def armstrong(n):
    temp= n
    digit= 0
    while temp > 0:
        temp= temp//10
        digit+=1
    temp= n
    total= 0
    while temp >0:
        d= temp%10
        p = 1
        for i in range(digit):
            p *= d
            
        total += p
        temp //= 10

    return total == n

n= int(input("enter a number: "))
print(armstrong(n))
```

Write a function that returns the    largest digit    in a number (without converting to string). 

```python
def large(n):
    if n < 10:
        return n
    else:
        last = n % 10
        return max(last, large(n // 10))
n = int(input("Enter a number: "))
print("Largest digit is:",large(n))
```

4. Write a function that takes a number and returns    True    if the number is palindrome and    
False otherwise (without using string conversion). 


```python
def pallin(num):
    num1=num
    rev=0
    while(num!=0):
        digit=num%10
        rev=rev*10+digit
        num=num//10
    if(rev==num1):
        return True
    else:
        return False
n=int(input("enter a number"))
print("the number",n ,"is pallindrome ?",pallin(n))
```

6. Write a function that rotates digits of a number    k times to the right   . 
Example: rotate(12345,2) → 45123 

```python
def rotate(n, k):
    s = str(n)
    k %= len(s)
    return int(s[-k:] + s[:-k])

print(rotate(12345, 2))
```

7. Write a function that returns the    factorial of a number using recursion    without using loops. 

```python
def factorial(n):
    if n <= 1:
        return 1
    
    return n * factorial(n - 1)

print(factorial(5))
```

8. Write a function that checks whether a number is a    Perfect Number   . 

```python
def per(num):
    sum=0
    perfect=False
    for i in range(1,num):
        if(num%i==0):
            sum=sum+i
        if(sum==num):
            perfect=True
    return perfect
n=int(input("Enter a number:"))
print(f"is {n} perfect number:",per(n))
```

```python
def countvowels(s):
    count = 0
    for ch in s.lower():
        if ch in "aeiouAEIOU":
            count += 1
    return count
s = input("enter a string: ")
print("vowel count:", countvowels(s))
```

11. Write a function that returns the    second largest element    in a list without using `sort()` or 
`max()`.

```python
def second_largest(numbers):
    largest = None
    second = None
    for value in numbers:
        if largest is None or value > largest:
            second = largest
            largest = value
        elif value != largest and (second is None or value > second):
            second = value
    return second
values = [10, 20, 40, 30, 50]
print("Second largest:", second_largest(values))
```

10. Write a function that returns the    LCM of two numbers    using the GCD function. 

```python
def gcd(a , b):
    while (a!=b):
        if a>b:
            a-=b
        else:
            b-=a
    return a

def Lcm (a,b):
    return a*b / gcd(a,b)

print(Lcm(3,4))
```

16. Write a function that returns the    sum of factorial of each digit    of a number. 

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
    
def sumOfDigits(num):
    if num == 0:
        return num
    return (factorial(num%10) + sumOfDigits(num//10))

print(sumOfDigits(123))
```

13. Write a function that removes    duplicate elements    from a list without using `set()`.

```python
def remove_duplicates(lst):
    unique_list = []
    i = 0
    while(i < len(lst)):
        item = lst[i]
        
        found = False
        j = 0
        while(j < len(unique_list)):
            if(unique_list[j] == item):
                found = True
            j = j + 1
            
        if(found == False):
            unique_list.append(item)
            
        i = i + 1
    return unique_list

nums = [1, 2, 3, 2, 1, 4, 5, 4]
t1 = remove_duplicates(nums)
print("List after removing duplicates:", t1)
```

```python
def char_freq(s):
    freq_dict = {}
    i = 0
    while(i < len(s)):
        ch = s[i]
        
        if(ch in freq_dict):
            freq_dict[ch] = freq_dict[ch] + 1
        else:
            freq_dict[ch] = 1
            
        i = i + 1
    return freq_dict

text = "Suman Nayak"
t1= char_freq(text)
print("Character frequencies:", t1)
```
