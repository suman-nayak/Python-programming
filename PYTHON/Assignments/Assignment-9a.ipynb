Q1. Define a class Point, which can be used to create two-dimensional points. Create 
three objects of Point class and check, these points are non collinear. 

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

def are_collinear(p1, p2, p3):
    area = (p1.x * (p2.y - p3.y)+ p2.x * (p3.y - p1.y)+ p3.x * (p1.y - p2.y))
    return area == 0


p1 = Point(0, 0)
p2 = Point(2, 3)
p3 = Point(4, 7)

if are_collinear(p1, p2, p3):
    print("The points are collinear.")
else:
    print("The points are non-collinear.")
```

Q2. Define a class Box, having basic data members and member function and create Box 
objects and compute the volume of the Box

```python
class Box:

    def __init__(self, length, breadth, height):
        self.length = length
        self.breadth = breadth
        self.height = height

    def volume(self):
        return self.length * self.breadth * self.height

    def display(self):
        print("Length = ",self.length)
        print("Breadth =",self.breadth)
        print("Height =",self.height)
        print("Volume =", self.volume())


box1 = Box(10, 5, 4)
box2 = Box(12, 8, 6)


print("Details of Box 1:")
box1.display()

print("\nDetails of Box 2:")
box2.display()
```

Q3. Define a class ComplexNumber, placed the necessary data members and member 
functions to perform basic Complex number operation.

```python
class ComplexNumber:
    def __init__(self, r, i):
        self.r = r
        self.i = i

    def add(self, c):
        return ComplexNumber(self.r + c.r, self.i + c.i)

    def sub(self, c):
        return ComplexNumber(self.r - c.r, self.i - c.i)

    def mul(self, c):
        r = self.r * c.r - self.i * c.i
        i = self.r * c.i + self.i * c.r
        return ComplexNumber(r, i)

    def div(self, c):
        d = c.r ** 2 + c.i ** 2
        r = (self.r * c.r + self.i * c.i) / d
        i = (self.i * c.r - self.r * c.i) / d
        return ComplexNumber(r, i)

    def show(self):
        print(self.r, "+", self.i, "i")

a = ComplexNumber(3, 4)
b = ComplexNumber(1, 2)

print("First number:")
a.show()

print("Second number:")
b.show()

print("Add:")
a.add(b).show()

print("Subtract:")
a.sub(b).show()

print("Multiply:")
a.mul(b).show()

print("Divide:")
a.div(b).show()
```

Q4. Define a class Time having data members hour, minute and second. The class 
should have a parameterized constructor to initialize its data members. It should also have 
methods displayTime() to display the time in HH:MM: SS format and addTime() to add 
two-time objects

```python
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s

    def displayTime(self):
        print(self.h, ":", self.m, ":", self.s)

    def addTime(self, t):
        s = self.s + t.s
        m = self.m + t.m
        h = self.h + t.h

        if s >= 60:
            m = m + s // 60
            s = s % 60

        if m >= 60:
            h = h + m // 60
            m = m % 60

        return Time(h, m, s)


t1 = Time(2, 45, 50)
t2 = Time(1, 20, 30)

print("First time:")
t1.displayTime()

print("Second time:")
t2.displayTime()

print("Added time:")
t3 = t1.addTime(t2)
t3.displayTime()
```

```python
class BankAccount:
    def __init__(self, accNo, name, bal, accType):
        self.accNo = accNo
        self.name = name
        self.bal = bal
        self.accType = accType

    def deposit(self, amount):
        self.bal = self.bal + amount
        print("Deposited:", amount)
        print("New Balance:", self.bal)

    def withdraw(self, amount):
        if amount > self.bal:
            print("Insufficient balance!")
        else:
            self.bal = self.bal - amount
            print("Withdrawn:", amount)
            print("New Balance:", self.bal)

    def displayAccountDetails(self):
        print("Account Number:", self.accNo)
        print("Account Holder Name:", self.name)
        print("Account Type:", self.accType)
        print("Account Balance:", self.bal)


acc1 = BankAccount(101, "Rahul", 5000, "Savings")

acc1.displayAccountDetails()

print()
acc1.deposit(2000)

print()
acc1.withdraw(1000)

print()
acc1.displayAccountDetails()
```

Q6. Define a class Stack, which has the member function to perform basic operation of 
stack. Demonstrate the basic operation. 

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)
        print(item, "pushed into stack")

    def pop(self):
        if self.isEmpty():
            print("Stack is empty. Cannot pop.")
        else:
            item = self.items.pop()
            print(item, "popped from stack")

    def peek(self):
        if self.isEmpty():
            print("Stack is empty.")
        else:
            print("Top element:", self.items[-1])

    def isEmpty(self):
        return len(self.items) == 0

    def size(self):
        print("Stack size:", len(self.items))

    def display(self):
        print("Stack:", self.items)


s = Stack()

s.push(10)
s.push(20)
s.push(30)

s.display()
s.peek()
s.size()

s.pop()
s.display()

s.pop()
s.pop()
s.pop()
```

```python
class Student:
    def __init__(self, name, roll, marks, branch, college):
        self.studName = name
        self.studRoll = roll
        self.studMarks = marks
        self.studBranch = branch
        self.studCollegeName = college

    def getAverage(self):
        return sum(self.studMarks) / len(self.studMarks)

    def dispStudentDetails(self):
        print("Student Name:", self.studName)
        print("Student Roll Number:", self.studRoll)
        print("Marks in 5 Subjects:", self.studMarks)
        print("Average Marks:", self.getAverage())
        print("Branch:", self.studBranch)
        print("College Name:", self.studCollegeName)


students = [
    Student("Rahul", 101, [85, 90, 78, 92, 88], "CSE", "ABC College"),
    Student("Priya", 102, [75, 80, 70, 85, 90], "IT", "ABC College"),
    Student("Amit", 103, [95, 92, 88, 90, 94], "CSE", "ABC College"),
    Student("Sneha", 104, [60, 70, 65, 75, 80], "ECE", "ABC College"),
    Student("Vikram", 105, [88, 84, 90, 86, 82], "ME", "ABC College"),
    Student("Neha", 106, [70, 75, 80, 85, 90], "CSE", "ABC College"),
    Student("Arjun", 107, [90, 85, 95, 80, 88], "IT", "ABC College"),
    Student("Kiran", 108, [65, 70, 75, 60, 80], "ECE", "ABC College"),
    Student("Divya", 109, [92, 88, 85, 90, 95], "CSE", "ABC College"),
    Student("Ravi", 110, [78, 82, 75, 80, 85], "ME", "ABC College")
]

topStudent = students[0]

for student in students:
    if student.getAverage() > topStudent.getAverage():
        topStudent = student

print("Student with Highest Average Marks:")
topStudent.dispStudentDetails()
```

```python
class Employee:
    def __init__(self, name, number, basic, da, hra):
        self.empName = name
        self.empNumber = number
        self.basicSal = basic
        self.DA = da
        self.HRA = hra

    def calGrossSal(self):
        return self.basicSal + self.DA + self.HRA

    def showEmpDetails(self):
        print("Employee Name:", self.empName)
        print("Employee Number:", self.empNumber)
        print("Basic Salary:", self.basicSal)
        print("DA:", self.DA)
        print("HRA:", self.HRA)
        print("Gross Salary:", self.calGrossSal())
        print()

employees = [
    Employee("Rahul", 101, 30000, 5000, 4000),
    Employee("Priya", 102, 40000, 8000, 6000),
    Employee("Amit", 103, 35000, 7000, 5000),
    Employee("Sneha", 104, 50000, 10000, 8000),
    Employee("Vikram", 105, 25000, 4000, 3000)
]

print("Employees with Gross Salary more than 50,000:")
print()

for emp in employees:
    if emp.calGrossSal() > 50000:
        emp.showEmpDetails()
```
