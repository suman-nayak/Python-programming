```python
phy = int(input("Enter physics mark: "))
che = int(input("Enter chemistry mark: "))
math = int(input("Enter math mark: "))
total = phy + che + math
percentage = (total/300)*100
if(percentage > 90):
    print("O grade")
elif(percentage > 80):
    print("E grade")
elif(percentage > 70):
    print("A Grade")
elif(percentage > 60):
    print("B grade")
elif(percentage > 50):
    print("C grade")
```
