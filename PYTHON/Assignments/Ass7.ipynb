```python
#Q3

months = {
    "January": 31,
    "February": 28,
    "March": 31,
    "April": 30,
    "May": 31,
    "June": 30,
    "July": 31,
    "August": 31,
    "September": 30,
    "October": 31,
    "November": 30,
    "December": 31
}

month = input("Enter a month name: ")

if month in months:
    print("Number of days in", month, "=", months[month])
else:
    print("Invalid month name")


print("\nMonths in alphabetical order:")

month_names = list(months.keys())
month_names.sort()

for month in month_names:
    print(month)

print("\nMonths with 31 days:")

for month in months:
    if months[month] == 31:
        print(month)


print("\nMonths sorted by number of days:")

month_days = list(months.items())

for i in range(len(month_days)):
    for j in range(i + 1, len(month_days)):
        if month_days[i][1] > month_days[j][1]:
            temp = month_days[i]
            month_days[i] = month_days[j]
            month_days[j] = temp

for month, days in month_days:
    print(month, ":", days)
```

```python
#Q4
x = {
    'k1': 'v1',
    'k2': 'v2',
    'k3': 'v3'
}
inverted_x = {}

for key in x:
    inverted_x[x[key]] = key

print("Original Dictionary:", x)
print("Inverted Dictionary:", inverted_x)
```

```python
#Q5
D1 = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

D2 = {
    'c': 50,
    'd': 60,
    'e': 70,
    'f': 80
}

overlapping_keys = []

for key in D1.keys():
    if key in D2.keys():
        overlapping_keys.append(key)

print("Overlapping keys:", overlapping_keys)
```

```python
#Q7
d1 = {
    1: 11,
    2: 12
}

d2 = {
    1: 11,
    2: 12,
    3: 13,
    4: 15
}

contained = True

for key in d1:
    if key not in d2 or d1[key] != d2[key]:
        contained = False
        break

if contained:
    print("d1 is contained in d2")
else:
    print("d1 is not contained in d2")
```

```python
#Q10
friends = {}

n = int(input("Enter the number of friends: "))

for i in range(n):
    name = input("Enter friend name: ")
    phone = input("Enter phone number: ")
    friends[name] = phone


print("\nFriends and Phone Numbers:")

for name in friends:
    print(name, ":", friends[name])


new_name = input("\nEnter name of new friend: ")
new_phone = input("Enter phone number: ")

friends[new_name] = new_phone

print("Modified Dictionary:")
print(friends)


delete_name = input("\nEnter the name of friend to delete: ")

if delete_name in friends:
    del friends[delete_name]
    print("Friend deleted successfully.")
else:
    print("Friend not found.")

print("Dictionary after deletion:")
print(friends)


modify_name = input("\nEnter the name of friend to modify: ")

if modify_name in friends:
    new_phone = input("Enter new phone number: ")
    friends[modify_name] = new_phone
    print("Phone number modified successfully.")
else:
    print("Friend not found.")

print("Dictionary after modification:")
print(friends)


check_name = input("\nEnter the name of friend to search: ")

if check_name in friends:
    print(check_name, "is present in the dictionary.")
else:
    print(check_name, "is not present in the dictionary.")


print("\nDictionary in sorted order of names:")

names = list(friends.keys())
names.sort()

for name in names:
    print(name, ":", friends[name])
```

```python
#Q11
paragraph = input("Enter a paragraph: ")

words = paragraph.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print("Word Frequency:")
print(frequency)
```

```python
#Q13
D1 = {'a': 1, 'b': 2, 'c': 1, 'd': 3, 'e': 2}

D2 = {}

for key in D1:
    value = D1[key]

    if value in D2:
        D2[value].append(key)
    else:
        D2[value] = [key]

print("Original Dictionary:", D1)
print("Inverted Dictionary:", D2)
```

```python
#Q14
string = input("Enter a string: ")

frequency = {}

for ch in string:
    if ch in frequency:
        frequency[ch] = frequency[ch] + 1
    else:
        frequency[ch] = 1

for ch in string:
    if frequency[ch] == 1:
        print("First non-repeating character:", ch)
        break
```

```python
#Q15
numbers = []

n = int(input("Enter the number of elements: "))

for i in range(n):
    num = int(input("Enter a number: "))
    numbers.append(num)

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print("Number Frequency:")
print(frequency)
```

```python
#Q16
words = ["cat", "dog", "elephant", "rat", "tiger"]

result = {}

for word in words:
    length = len(word)

    if length in result:
        result[length].append(word)
    else:
        result[length] = [word]

print("Dictionary:", result)
```

```python
#Q17
D = {'a': 40, 'b': 10, 'c': 30, 'd': 20}

items = list(D.items())

for i in range(len(items)):
    for j in range(i + 1, len(items)):
        if items[i][1] > items[j][1]:
            temp = items[i]
            items[i] = items[j]
            items[j] = temp

sorted_D = {}

for item in items:
    sorted_D[item[0]] = item[1]

print("Original Dictionary:", D)
print("Dictionary sorted by values:", sorted_D)
```
