1. Create a tuple with different data and create an empty tuple using built-in  
function.  

```python
t1 = (1, True, 2.4, "Silicon")
empt = tuple()

print("The mixed tuple is: ", t1)
print("Empty tuple is: ", empt)
```

2. Print the 3rd element from beginning and 4th element from end of a tuple. 

```python
a = (10, 20, 30.5, True, 50, 60, 70)
print("3rd element from beginning: ",a[2])
print("4th element from end:", a[-4])
```

3. Count the number of items present in a tuple and then add two new items   
into it (one item using concatenation operator and another item using built-in  
function).

```python
tup = (10, 20, 30)

count = len(tup)
print("Number of items in the tuple:", count)

tup = tup + (40,)
print("After concatenation:", tup)

temp = list(tup)
temp.append(50)
tup = tuple(temp)
print("After using built-in functions:", tup)
```

4. Find the number of occurrences for any repeated item present in a tuple.

```python
t = (10, 20, 30, 20, 40, 20, 50, 30)
n = int(input("Enter the item: "))

print("The occurrence is: ", t.count(n))
```

5. Display the elements present in a tuple which are at the even position.
