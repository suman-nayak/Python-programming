3.	Create a dictionary whose keys are month names and whose values are the number of days in the corresponding months.
(a)	Ask the user to enter a month name and use the dictionary to tell how many days are in the month.
(b)	Print out all of the keys in alphabetical order.
(c)	Print out all of the months with 31 days.
(d)	Print out the (key-value) pairs sorted by the number of days in each month.


```python
months = {
    "January": 31,
    "February": 28,
    "March": 31,
    "April": 30,
    "May": 31,
    "June": 30,
    "July": 31,
    "August": 31,
    "September": 30,
    "October": 31,
    "November": 30,
    "December": 31
}

month = input("Enter a month name: ")

if month in months:
    print("Number of days in", month, "=", months[month])
else:
    print("Invalid month name")


print("\nMonths in alphabetical order:")

month_names = list(months.keys())
month_names.sort()

for month in month_names:
    print(month)

print("\nMonths with 31 days:")

for month in months:
    if months[month] == 31:
        print(month)


print("\nMonths sorted by number of days:")

month_days = list(months.items())

for i in range(len(month_days)):
    for j in range(i + 1, len(month_days)):
        if month_days[i][1] > month_days[j][1]:
            temp = month_days[i]
            month_days[i] = month_days[j]
            month_days[j] = temp

for month, days in month_days:
    print(month, ":", days)
```

4.	Given the dictionary x = {'k1':'v1', 'k2':'v2', 'k3':'v3'}, create a dictionary with the opposite mapping, i.e., write a program to create the dictionary as :
inverted_x = {'v1': 'k1' , 'v2' :'k2' , 'v3':'k3'}


```python
x = {
    'k1': 'v1',
    'k2': 'v2',
    'k3': 'v3'
}
inverted_x = {}

for key in x:
    inverted_x[x[key]] = key

print("Original Dictionary:", x)
print("Inverted Dictionary:", inverted_x)
```

5.	Given two dictionaries say D1 and D2. Write a program that lists the overlapping keys of the two dictionaries, i.e., if a key of D1 is also a key of D2, then list it.

```python
D1 = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

D2 = {
    'c': 50,
    'd': 60,
    'e': 70,
    'f': 80
}

overlapping_keys = []

for key in D1.keys():
    if key in D2.keys():
        overlapping_keys.append(key)

print("Overlapping keys:", overlapping_keys)
```

6.	Write a program that checks if two same values in a dictionary have different keys. That is, for dictionary D1 = { 'a' : 10, 'b': 20, 'c' : 10}, the program should print 2 keys have same values and for dictionary D2 = {'a' : 10, 'b' : 20, 'c' : 30} , the program should print No keys have same values.

```python
D1 = {
    'a': 10,
    'b': 20,
    'c': 10
}

values = []

for key in D1:
    if D1[key] in values:
        print("2 keys have same values")
        break
    else:
        values.append(D1[key])
else:
    print("No keys have same values")
```

7.	Write a program to check if a dictionary is contained in another dictionary e.g., if
d1 = {1:11, 2:12}
d2 = {1:11, 2:12, 3:13, 4:15}
then d1 is contained in d2.


```python
d1 = {
    1: 11,
    2: 12
}

d2 = {
    1: 11,
    2: 12,
    3: 13,
    4: 15
}

contained = True

for key in d1:
    if key not in d2 or d1[key] != d2[key]:
        contained = False
        break

if contained:
    print("d1 is contained in d2")
else:
    print("d1 is not contained in d2")
```

8.	A dictionary D1 has values in the form of lists of numbers. Write a program to create a new dictionary D2 having same keys as D1 but values as the sum of the list elements e.g.,
D1 = {'A' : [1, 2, 3] , 'B' : [4, 5, 6]}
then
D2 is {'A' :6, 'B' : 15}


```python
D1 = {
    'A': [1, 2, 3],
    'B': [4, 5, 6]
}

D2 = {}

for key in D1:
    D2[key] = sum(D1[key])

print("D1 =", D1)
print("D2 =", D2)
```

10.	Write a program to input your friends’ names and their Phone Numbers and store them in the
dictionary as the key-value pair. Perform the following operations on the dictionary:
a) Display the name and phone number of all your friends
b) Add a new key-value pair in this dictionary and display the modified dictionary
c) Delete a particular friend from the dictionary
d) Modify the phone number of an existing friend
e) Check if a friend is present in the dictionary or not
f) Display the dictionary in sorted order of names


```python
friends = {}

n = int(input("Enter the number of friends: "))

for i in range(n):
    name = input("Enter friend name: ")
    phone = input("Enter phone number: ")
    friends[name] = phone


print("\nFriends and Phone Numbers:")

for name in friends:
    print(name, ":", friends[name])


new_name = input("\nEnter name of new friend: ")
new_phone = input("Enter phone number: ")

friends[new_name] = new_phone

print("Modified Dictionary:")
print(friends)


delete_name = input("\nEnter the name of friend to delete: ")

if delete_name in friends:
    del friends[delete_name]
    print("Friend deleted successfully.")
else:
    print("Friend not found.")

print("Dictionary after deletion:")
print(friends)


modify_name = input("\nEnter the name of friend to modify: ")

if modify_name in friends:
    new_phone = input("Enter new phone number: ")
    friends[modify_name] = new_phone
    print("Phone number modified successfully.")
else:
    print("Friend not found.")

print("Dictionary after modification:")
print(friends)


check_name = input("\nEnter the name of friend to search: ")

if check_name in friends:
    print(check_name, "is present in the dictionary.")
else:
    print(check_name, "is not present in the dictionary.")


print("\nDictionary in sorted order of names:")

names = list(friends.keys())
names.sort()

for name in names:
    print(name, ":", friends[name])
```

11.	Word Frequency in a Paragraph

```python
paragraph = input("Enter a paragraph: ")

words = paragraph.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print("Word Frequency:")
print(frequency)
```

12.	Find Common Keys in Two Dictionaries

```python
D1 = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
D2 = {'c': 50, 'd': 60, 'e': 70, 'f': 80}

common_keys = []

for key in D1:
    if key in D2:
        common_keys.append(key)

print("Common keys:", common_keys)
```
