```python
unit = float(input("Enter unit:"))
if (unit < 50):
    bill= 0.50 * unit
elif (unit < 150):
    bill= 0.50 * unit + 0.75 * unit
elif (unit < 250):
    bill = 0.50 * unit + 0.75 * unit + 1.20 * unit
elif(unit > 250):
    bill = 0.50 * unit + 0.75 * unit + 1.20 * unit + unit * 1.50

print("The Electricity bill is:",bill + bill*20/100)
```
