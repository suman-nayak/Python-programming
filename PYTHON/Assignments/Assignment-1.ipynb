```python
#20.	Accept two numbers and perform the all logical, bitwise operation and display the result.
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("\n--- Logical Operations ---")
print("a and b =", a and b)
print("a or b  =", a or b)
print("not a   =", not a)
print("not b   =", not b)

print("\n--- Bitwise Operations ---")
print("a & b  =", a & b)    
print("a | b  =", a | b)    
print("a ^ b  =", a ^ b)    
print("~a     =", ~a)       
print("~b     =", ~b)
print("a << 1 =", a << 1)   
print("b << 1 =", b << 1)
print("a >> 1 =", a >> 1)   
print("b >> 1 =", b >> 1)
```
