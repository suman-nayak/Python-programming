```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("\n--- Logical Operations ---")
print("a and b =", a and b)
print("a or b  =", a or b)
print("not a   =", not a)
print("not b   =", not b)

print("\n--- Bitwise Operations ---")
print("a & b  =", a & b)    # AND
print("a | b  =", a | b)    # OR
print("a ^ b  =", a ^ b)    # XOR
print("~a     =", ~a)       # NOT
print("~b     =", ~b)
print("a << 1 =", a << 1)   # Left shift
print("b << 1 =", b << 1)
print("a >> 1 =", a >> 1)   # Right shift
print("b >> 1 =", b >> 1)
```
