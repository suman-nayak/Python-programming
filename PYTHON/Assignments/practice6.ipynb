```python
num = int(input("Enter the number of employees : "))
count=1
employee = {}

while count <= num:
    name = input("Enter the name of the Employee: ")
    salary = int(input("Enter the salary: "))
    employee[name] = salary
    count=count+1
print(employee) 
```
