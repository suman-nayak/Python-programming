1. Create a Class for a Student 
Write a class Student with attributes: 
• name 
• age 
• marks 
Create a method display() that prints the student details.  

```python
class Student:
    def __init__(self, name, age, mark):
        self.name = name
        self.age = age
        self.mark = mark
    
    def display(self):
        print("Name:", self.name )
        print("Age:", self.age)
        print("Mark:", self.mark)
    
s1 = Student("suman", 23, 98)
s1.display()
```

2. Rectangle Area 
Create a class Rectangle with: 
• length 
• width 
Add methods: 
• area() → returns area 
• perimeter() → returns perimeter 

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2*(self.length * self.width)
    
    
r1 = Rectangle(23, 32)
print(r1.area())
print(r1.perimeter())
```
