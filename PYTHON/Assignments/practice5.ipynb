Write a program that rotates the element of a list so that the

element at the first index moves to the second index, the element

in the second index moves to the third index, etc., and the

element in the last index moves to the first index.

```python
list = [10, 20, 30, 40, 50]

if len(list) > 1:
    first = list[0]    
    for i in range(len(list) - 1):
        list[i] = list[i + 1]
    list[-1] = first

print(list)
```
